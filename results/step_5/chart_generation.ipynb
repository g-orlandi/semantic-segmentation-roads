```python
import matplotlib.pyplot as plt
import numpy as np

# --- I TUOI DATI ORIGINALI ---
# Epoche: 0, 1, 2, ..., 14
epoche = list(range(15))

# Liste con i dati riscalati (divisi per 100)
miou_head = [0.60, 0.666, 0.70, 0.706, 0.711, 0.718, 0.711, 0.712, 0.718, 0.719, 0.709, 0.713, 0.706, 0.712, 0.705]
miou_3comp = [0.422, 0.497, 0.513, 0.542, 0.560, 0.575, 0.563, 0.572, 0.576, 0.576, 0.572, 0.567, 0.570, 0.573, 0.573]
miou_4comp = [0.609, 0.669, 0.686, 0.710, 0.720, 0.723, 0.698, 0.711, 0.705, 0.700, 0.704, 0.707, 0.710, 0.713, 0.714]

# --- SIMULAZIONE BARRE DI ERRORE (come nell'esempio condiviso) ---
# Assegniamo un errore std di ~0.015 per tutte le epoche, per replicare lo stile dell'esempio
# In un caso reale, avresti dati diversi per ogni modello e ogni epoca
err_std = 0.015 # ampiezza della barra di errore (simile all'esempio)
err_head = [err_std] * 15
err_3comp = [err_std] * 15
err_4comp = [err_std] * 15

# --- CREAZIONE DEL GRAFICO ---
plt.figure(figsize=(10, 7), dpi=100) # Dimensioni e DPI simili all'esempio

# Stile di base pulito
plt.rc('font', size=12) # Dimensione font simile

# Usiamo errorbar per disegnare le linee con i punti e le barre di errore
# markeredgecolor='black' aggiunge un bordo ai punti, come nell'esempio
plt.errorbar(epoche, miou_4comp, yerr=err_4comp, fmt='o-', color='#e41a1c', label='4 Componenti (ip.)', capsize=5, markeredgecolor='black', markersize=8)
plt.errorbar(epoche, miou_head, yerr=err_head, fmt='o-', color='#377eb8', label='Solo Head', capsize=5, markeredgecolor='black', markersize=8)
plt.errorbar(epoche, miou_3comp, yerr=err_3comp, fmt='o-', color='#4daf4a', label='3 Componenti', capsize=5, markeredgecolor='black', markersize=8)

# --- TITOLO E ETICHETTE ASSI ---
plt.title('Confronto mIoU per diverse strategie di Fine-Tuning EoMT')
plt.xlabel('Epoca')
plt.ylabel('mIoU')

# --- CONFIGURAZIONE ASSI E GRIGLIA ---
plt.xticks(range(0, 15, 2)) # Tick ogni 2 epoche (0, 2, 4...)
# Un intervallo di y che mostri bene i dati, e che sia simile all'esempio (0.4 - 0.8)
plt.ylim(0.4, 0.8) # Modifica in base al tuo mIoU massimo

# Griglia grigia leggera come nell'esempio
plt.grid(True, which='both', color='#D3D3D3', linestyle='-', alpha=0.7)

# --- LEGENDA ---
# Posizione in alto a destra, sfondo bianco e trasparente come nell'esempio
legend = plt.legend(loc='upper right', frameon=True, fontsize=11)
frame = legend.get_frame()
frame.set_facecolor('white')
frame.set_alpha(0.8)
frame.set_edgecolor('grey')

# --- SALVATAGGIO ---
# Salva come PDF per la massima qualità vettoriale su Overleaf
# Per visualizzarlo e basta, commenta la riga sotto e scommenta plt.show()
plt.savefig('confronto_finetuning_miou.pdf', format='pdf', bbox_inches='tight')
# plt.show()

print("File 'confronto_finetuning_miou.pdf' generato con successo.")
```
